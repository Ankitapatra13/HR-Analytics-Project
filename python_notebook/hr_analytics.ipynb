```python
import pandas as pd
```

```python
# import dataset
df = pd.read_csv("Uncleaned_employees_final_dataset (1).csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# null check
df.isnull().sum()
```

```python
# filling the null values of "education" feature with "Unknown"

df["education"] = df["education"].fillna("Unknown")
```

```python
# checking if duplicates exists

df.duplicated()
```

```python
# dropping the duplicates

df = df.drop_duplicates()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["gender"].unique()
```

```python
# gender mapping in gender column
gender_mapping = {"m" : "male", "f" : "female"}

df["gender"] = df["gender"].map(gender_mapping)
```

```python
df.head()
```

```python
## feature engineering
## converting discrete age numbers into age brackets in a new feature "age_group".

bin_labels = ["Young adult", "Adult", "Middle-aged", "Senior citizen"]
bin_edges = [18,25,40,55,65]
df["age_group"] = pd.cut(df["age"], bins = bin_edges, labels = bin_labels)
```

```python
df.sample(20)
```

```python
df.isnull().sum()
```

```python
# filling the null values of "previous_year_rating" with the mode vale within their department 

df["previous_year_rating"] = df["previous_year_rating"].fillna(df.groupby("department")["previous_year_rating"].transform(lambda x : x.mode()[0]))
```

```python
df.isnull().sum()
```

```python
# Renaming the column "KPIs_met_more_than_80" to lowercase and only characters 

df = df.rename(columns={"KPIs_met_more_than_80" : "kpis_met_more_than_eighty"})
```

```python
df.columns
```

```python
df
```

```python
# feature engineering
# creating a new feature work experience from the feature length of service

bin_labels = ["Entry-level(0 to 2 yrs)", "Mid-level(3 to 7 yrs)", "Senior-level(7 to 12 yrs)", "Executive-level(12+ yrs)"]
bin_edges = [0,2,7,12,40]
df["work_experience"] = pd.cut(df["length_of_service"] , bins = bin_edges , labels = bin_labels)   
```
